```python
##Cargar archivo
import pandas as pd
import matplotlib.pyplot as plt
df1 = pd.read_csv("Datos_Facturación.csv")
df1
```

**Filtros**

```python
## Ajustar máximo de filas
pd.options.display.max_rows = 50
```

```python
## 1-CVE_CLPV      (valores: 1000 a 2000)
filtro_CVE_CLPV = df1[(df1['CVE_CLPV'] >= 1000) & (df1['CVE_CLPV'] <= 2000)]
filtro_CVE_CLPV
filtro_CVE_CLPV.to_csv("filtro_CVE_CLPV.csv")
```

```python
## 2. 2-CVE_VEND     (Todas las claves excepto “5” y “4”)
filtro_CVE_VEND_= df1[~(df1["CVE_VEND"] == 5) & (df1["CVE_VEND"] == 4)]
filtro_CVE_VEND_
filtro_CVE_VEND_.to_csv("filtro_CVE_VEND_.csv")
```

```python
## 3-FECHA_ENT   (Fechas de “28/02/2022”)
filtro_FECHA_ENT = df1[df1['FECHA_ENT'] == "2022-02-28 00:00:00"]
filtro_FECHA_ENT
filtro_FECHA_ENT.to_csv("filtro_FECHA_ENT.csv")
```

```python
## 4-CAN_TOT       (Cantidades menores a 5951.7)   o  STATUS (“E”)
filtro4=df1[(df1["CAN_TOT"] < 5951.7) | (df1["STATUS"] == "E")]
filtro4
filtro4.to_csv("filtro4.csv")
```

```python
## 5-Solo las columnas: CVE_DOC, FECHA_ENT, FECHA_VEN y CAN_TOT
filtro5 = df1[["CVE_DOC", "FECHA_ENT", "FECHA_VEN", "CAN_TOT"]]
filtro5
filtro5.to_csv("filtro5.csv")
```

```python
## 6-Solo las filas de 7001-7099
filtro6 = df1.iloc[7000:7100]
filtro6
filtro6.to_csv("filtro6.csv")
```

```python
## 7-Index= CVE_VEND (valores: 1, 2)  (columna: FECHAELAB)
filtro7_1 = df1[df1["CVE_VEND"].isin([1, 2])]
filtro7_2 = filtro7_1[["CVE_VEND", "FECHAELAB"]]
filtro7_2.to_csv("filtro7_2.csv")
```
